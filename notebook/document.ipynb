```python
import os
os.makedirs("../data/text_files",exist_ok=True)
```

```python
sample_texts={
     "../data/text_files/AI_intro.txt":"""Artificial Intelligence is the simulation of human intelligence by machines.
Machine Learning is a subset of AI that allows systems to learn from data.
Deep Learning uses neural networks with multiple layers to process complex patterns.
Large Language Models like GPT and Claude are trained on massive text datasets.
AI is being used in healthcare, finance, automation, and many other industries.
Natural Language Processing enables machines to understand and generate human language.""",
"../data/text_files/python_facts.txt":"""Python is a high-level, interpreted programming language created by Guido van Rossum.
It was first released in 1991 and has become one of the most popular languages worldwide.
Python is widely used in data science, machine learning, web development, and automation.
Libraries like Pandas, NumPy, and Matplotlib make Python powerful for data analysis.
Python uses indentation to define code blocks instead of curly braces.
Frameworks like Django and Flask are used to build web applications with Python."""
}

for filepath,content in sample_texts.items():
    with open(filepath,'w',encoding="utf-8") as f:
        f.write(content)
print("Done!!!!")
```

```python
from langchain_community.document_loaders import TextLoader

loader=TextLoader("../data/text_files/AI_intro.txt",encoding="utf-8")
doc=loader.load()
print(doc)
```

```python
from langchain_community.document_loaders import DirectoryLoader

dir_loader=DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt",
    loader_cls= TextLoader,
    loader_kwargs={'encoding':'utf-8'},
    show_progress=False
)

documents=dir_loader.load()
documents
```

```python
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader, DirectoryLoader

dir_loader=DirectoryLoader(
    "../data/pdf_files",
    glob="**/*.pdf",
    loader_cls= PyMuPDFLoader,
    show_progress=False
)

documents=dir_loader.load()
documents
```
